*Exercise 3:* Modify Exercise 2 so that it takes a matrix in the format from Exercise 1 and draws a tic‑tac‑toe board with `"X"`s and `"O"`s.

````python
# Write your solution here
def draw_board(board):
    n = len(board)
    line = "--- " * n

    for row in board:
        print(line)

        for value in row:
            if value == 1:
                print("| X ", end="")
            elif value == 2:
                print("| O ", end="")
            else:
                print("|   ", end="")

        print("|")

    print(line)
````

*Exercise 4:* Write a function that takes an `n×n` matrix representing a tic‑tac‑toe game and returns one of the following values:

- `-1` if the game is **incomplete** (still empty spaces and no winner)
- `0` if the game is a **draw**
- `1` if **player 1** (`"X"`) has won
- `2` if **player 2** (`"O"`) has won

Here are some example inputs you can use to test your code:


````python
# Write your solution here
def game_status(board):
    n = len(board)
    for row in board:
        if row.count(1) == n:
            return 1
        if row.count(2) == n:
            return 2
    for col in range(n):
        column = [board[row][col] for row in range(n)]
        if column.count(1) == n:
            return 1
        if column.count(2) == n:
            return 2
    diagonal = [board[i][i] for i in range(n)]
    if diagonal.count(1) == n:
        return 1
    if diagonal.count(2) == n:
        return 2
    diagonal = [board[i][n - 1 - i] for i in range(n)]
    if diagonal.count(1) == n:
        return 1
    if diagonal.count(2) == n:
        return 2
    for row in board:
        if 0 in row:
            return -1
    return 0
````

````python
# Test your solution here
board = [
    [1, 1, 1],
    [2, 2, 0],
    [0, 0, 0]
]

print(game_status(board))
print(game_status([
    [1, 0, 2],
    [0, 1, 0],
    [2, 0, 0]
]))
print(game_status([
    [1, 2, 1],
    [2, 1, 2],
    [2, 1, 2]
]))
print(game_status([
    [2, 1, 0],
    [2, 1, 0],
    [2, 0, 1]
]))
````

*Exercise 5:* Write a function that takes a game board, a player number, and `(row, col)` coordinates and places the correct mark (`"X"` or `"O"`) in that location.

Requirements:

- Only allow placing a mark in a previously empty location.
- Return `True` if the move was successful, and `False` otherwise.


````python
# Write your solution here
def make_move(board, player, row, col):
    if board[row][col] != 0:
        return False

    board[row][col] = player
    return True
````

*Exercise 6:* Modify Exercise 3 to show **row and column labels** so that players can specify locations like `"A2"` or `"C1"`.

````python
# Write your solution here
def draw_board(board):
    n = len(board)
    line = "   " + "--- ".join(["---"] * n)

    # Column labels
    print("    " + "   ".join(str(i + 1) for i in range(n)))

    for i in range(n):
        print("  " + line[3:])
        
        row = chr(65 + i) + " "
        
        for value in board[i]:
            if value == 1:
                row += "| X "
            elif value == 2:
                row += "| O "
            else:
                row += "|   "

        row += "|"
        print(row)

    print("  " + line[3:])
````

*Exercise 7:* Write a function that takes a board, a player number, and a location string (as in Exercise 6), then uses your function from Exercise 5 to update the board.

````python
# Write your solution here
def make_move_location(board, player, location):
    row = ord(location[0].upper()) - ord("A")
    col = int(location[1:]) - 1

    return make_move(board, player, row, col)
````

*Exercise 8:* Write a function that is called with a board and player number, takes input from the player using Python's `input()`, and modifies the board using your function from Exercise 7.

Keep asking for input until the player enters a valid location that results in a valid move.


````python
# Write your solution here
def player_move(board, player):
    while True:
        location = input("Enter your move (example: A2): ")

        try:
            if make_move_location(board, player, location):
                return
            else:
                print("That location is already occupied. Try again.")
        except (ValueError, IndexError):
            print("Invalid location. Try again.")
````

````python
# Test your solution here
board = create_board(3)

player_move(board, 1)

draw_board(board)
````

*Exercise 9:* Use all of the previous exercises to implement a full tic‑tac‑toe game:

- draw the board,
- repeatedly ask two players for a location,
- apply valid moves,
- check the game status until a player wins or the game is a draw.


````python
# Write yourrr solution here
def play_game(n):
    board = create_board(n)
    player = 1

    while True:
        draw_board(board)

        print("Player", player, "turn")
        player_move(board, player)

        status = game_status(board)

        if status == 1:
            draw_board(board)
            print("Player 1 (X) wins!")
            break
        elif status == 2:
            draw_board(board)
            print("Player 2 (O) wins!")
            break
        elif status == 0:
            draw_board(board)
            print("It's a draw!")
            break

        if player == 1:
            player = 2
        else:
            player = 1
````

*Exercise 11:* Develop a version of the game where one player is the computer.

Note: you do **not** need an extensive search for the best move. For example, you can have the computer:
- block obvious losses
- otherwise try to create a winning row/column/diagonal


````python
# Write your solution here
def computer_move(board, player):
    n = len(board)
    opponent = 3 - player

    for row in range(n):
        for col in range(n):
            if board[row][col] == 0:
                board[row][col] = player

                if game_status(board) == player:
                    return

                board[row][col] = 0

    for row in range(n):
        for col in range(n):
            if board[row][col] == 0:
                board[row][col] = opponent

                if game_status(board) == opponent:
                    board[row][col] = player
                    return

                board[row][col] = 0

    for row in range(n):
        for col in range(n):
            if board[row][col] == 0:
                board[row][col] = player
                return
````

````python
# Test your solution here
board = create_board(3)

make_move(board, 1, 0, 0)
make_move(board, 1, 1, 1)

computer_move(board, 2)

draw_board(board)
````

*Exercise 12:* Develop a version of the game where one player is the computer. This time, write a computer player using exhaustive search with a max depth parameter, similar to lecture.

````python
# Write your solution here
def computer_move(board, player, max_depth):
    opponent = 3 - player

    def search(depth, current_player):
        status = game_status(board)

        if status == player:
            return 1
        if status == opponent:
            return -1
        if status == 0:
            return 0
        if depth == 0:
            return 0

        moves = []

        for row in range(len(board)):
            for col in range(len(board)):
                if board[row][col] == 0:
                    board[row][col] = current_player
                    score = search(depth - 1, 3 - current_player)
                    board[row][col] = 0
                    moves.append(score)

        if current_player == player:
            return max(moves)
        else:
            return min(moves)

    best_score = -2
    best_move = None

    for row in range(len(board)):
        for col in range(len(board)):
            if board[row][col] == 0:
                board[row][col] = player
                score = search(max_depth - 1, opponent)
                board[row][col] = 0

                if score > best_score:
                    best_score = score
                    best_move = (row, col)

    if best_move is not None:
        row, col = best_move
        board[row][col] = player
````

````python
# Test your solution here
board = create_board(3)
make_move(board, 1, 0, 0)
make_move(board, 2, 1, 1)
computer_move(board, 2, 4)
draw_board(board)
````

*Exercise 13:* Make the 2 computer players play each-other for 10 games on a 3x3, then 4x4, then 5x5 grid. Set the max depth so that the games only take seconds. Measure the "smarter" player's win rate for each grid.

````python
# Write your solution here
def computer_game(n, depth1, depth2):
    board = create_board(n)
    player = 1

    while True:
        if player == 1:
            computer_move(board, 1, depth1)
        else:
            computer_move(board, 2, depth2)

        status = game_status(board)

        if status != -1:
            return status

        player = 3 - player


def run_games(n, depth1, depth2, games=10):
    player1_wins = 0
    player2_wins = 0
    draws = 0

    for i in range(games):
        result = computer_game(n, depth1, depth2)

        if result == 1:
            player1_wins += 1
        elif result == 2:
            player2_wins += 1
        else:
            draws += 1
    win_rate = player1_wins / games * 100
    print("Board:", n, "x", n)
    print("Computer 1 wins:", player1_wins)
    print("Computer 2 wins:", player2_wins)
    print("Draws:", draws)
    print("Computer 1 win rate:", win_rate, "%")
````

````python
# Test your solution here
run_games(3, 5, 3)
run_games(4, 3, 2)
run_games(5, 2, 1)
````
